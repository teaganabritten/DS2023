```python
import dash, plotly.express as px
from dash import dcc, html,Output, Input, Dash

gap =px.data.gapminder()
```

```python
app = Dash()

app.layout=html.Div([
    html.H1('Gapminder Dashboard'),
    dcc.RadioItems(id="vars",
                   options=[{'label':'Life Expectancy', 'value':'lifeExp'},
                            {'label':'Population', 'value':'pop'},
                            {'label': 'GDP per Capita', 'value':'gdpPercap'}],
                   value = 'gdpPercap'),
    dcc.Slider(id='year_slider',
               min=gap['year'].min(),
               max=gap['year'].max(),
               value=gap['year'].min(),
               marks={str(year): str(year) for year in gap['year'].unique()},
               step=5),
    dcc.Graph(id='map')
])
```

```python
@app.callback(
    Output('map', 'figure'),
    [Input('vars', 'value'),
     Input('year_slider', 'value')])

def update_map(variables,year_selected):
  gap_filtered=gap[gap['year']==year_selected]
  fig=px.choropleth(gap_filtered, locations='iso_alpha', color=variables,hover_name='country',
                    color_continuous_scale='viridis')
  return fig
```

```python
if __name__ == '__main__':
  app.run(debug=True)
```
