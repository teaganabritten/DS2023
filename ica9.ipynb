```python
import dash
from dash import Dash, html, dcc, Input, Output
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from io import BytesIO
import base64
import numpy as np
import pandas as pd
import plotly.express as px


df = pd.read_csv('https://raw.githubusercontent.com/teaganabritten/DS2023/refs/heads/main/Guns.csv')
app2 = Dash(__name__)

# Assuming 'law' column exists in your CSV with 'yes' or 'no' values
available_laws = df['law'].unique().tolist()
crime_options = [{'label': 'Violent Crime Rate', 'value': 'violent'},
                 {'label': 'Murder Rate', 'value': 'murder'},
                 {'label': 'Robbery Rate', 'value': 'robbery'}]

app2.layout = html.Div([
    html.H1("Violent Crime Dashboard"),
    html.Div([
        html.Label("Select Carry Law Status (Yes or No):"),
        dcc.Dropdown(
            id='law_dropdown',
            options=[{'label': law, 'value': law} for law in available_laws],
            value=available_laws[0] if available_laws else None,  # Set initial value
            clearable=False
        )
    ], style={'width': '48%', 'display': 'inline-block'}),
    html.Div([
        html.Label("Select Crime Type:"),
        dcc.Dropdown(
            id='crime_dropdown',
            options=crime_options,
            value='violent',  # Set initial crime type
            clearable=False
        )
    ], style={'width': '48%', 'display': 'inline-block', 'float': 'right'}),
    dcc.Slider(id='year_slider',
               min=df['year'].min(),
               max=df['year'].max(),
               value=df['year'].min(),
               marks={str(year): str(year) for year in df['year'].unique()},
               step=None),
    dcc.Graph(id='map')
])

@app2.callback(
    Output('map', 'figure'),
    [Input('crime_dropdown', 'value'),
     Input('year_slider', 'value'),
     Input('law_dropdown', 'value')]
)
def update_map(selected_crime, year_selected, selected_law):
    print(f"Callback triggered with crime: {selected_crime}, year: {year_selected}, law: {selected_law}")
    filtered_df = df[(df['law'] == selected_law) & (df['year'] == year_selected)]
    fig = px.choropleth(filtered_df, locations='state', locationmode='USA-states', scope='usa', color=selected_crime, hover_name='state', color_continuous_scale='viridis')
    return fig

if __name__ == '__main__':
    app2.run()
```
